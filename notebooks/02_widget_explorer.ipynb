# Neuro-Topology Widget Explorer
Interactive scrubbing of persistence diagrams over time.

```python
import numpy as np
from topoconscious.topology import PersistenceEngine
from topoconscious.visualization import TopoVisualizer

rng = np.random.default_rng(0)
ts = rng.standard_normal((300, 90))
engine = PersistenceEngine(max_dim=2, n_landmarks=50)
windows = [ts[t:t+30] for t in range(0, 270, 5)]
diagrams = [engine.compute(w) for w in windows]
```

```python
viz = TopoVisualizer()
widget = viz.persistence_widget(diagrams, tr=2.0)
from IPython.display import display
display(widget)
```
